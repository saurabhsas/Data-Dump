```python
import numpy as np
import pandas as pd
```

```python
array_df = np.column_stack((products, units_sold, regions, sales, cogs, profit, discounts))
print(array_df)
```

#### Phase 1: Setup & Conversion
* Using the lists created in the previous exercise, your first task is to move from a NumPy array to a Pandas DataFrame.

```python
df = pd.DataFrame(array_df, columns=['Product','Units_Sold','Region','Sales','COGS','Profit','Discount'])
df.head()
```

```python
numeric_cols = ['Units_Sold','Sales','COGS','Profit','Discount']
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric)

df.head()
```

#### Phase 3: Slicing & Dicing (loc vs iloc)
* Participants often get confused between label-based and integer-based indexing. Practice both:
* Integer Slicing (iloc): Extract the first 10 rows and the first 3 columns.
* Label Slicing (loc): Extract rows 5 through 15, but only the Product and Profit columns.
* Specific Selection: Use loc to find the Sales and Region for the 25th sample in the list.

```python
#Integer Slicing (iloc): Extract the first 10 rows and the first 3 columns.
df.iloc[:10,:3]
```

```python
# Label Slicing (loc): Extract rows 5 through 15, but only the Product and Profit columns.
df.loc[5:15,['Product', 'Profit']]
```

```python
# Specific Selection: Use loc to find the Sales and Region for the 25th sample in the list.
df.loc[24, ["Sales", "Region"]]
```

#### Phase 4: Advanced Querying
* The .query() function allows you to write filters using a syntax similar to SQL.
* High Volume: Use .query() to find all rows where Units_Sold > 15.
* Loss Leaders: Use .query() to find any rows where Profit < 0.
* Complex Logic: Find all products in the "North" region that had a Discount greater than 0.

```python
# High Volume: Use .query() to find all rows where Units_Sold > 15.
df.query('Units_Sold > 15')
```

```python
# Loss Leaders: Use .query() to find any rows where Profit < 0.
df.query('Profit < 0')
```

```python
# Complex Logic: Find all products in the "North" region that had a Discount greater than 0.
df.query('Region == "North" and Discount > 0')
```
